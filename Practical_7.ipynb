```python
# =====================================================================
# STEP 1: INSTALL PYSPARK & INITIALIZE SPARK SESSION
# =====================================================================
print("Installing PySpark...")
!pip install pyspark --quiet

from pyspark.sql import SparkSession
from pyspark.sql.functions import col, avg, sum, count, when, round

# Initialize Spark Session
spark = SparkSession.builder \
    .appName("Practical_7_PySpark_Lab") \
    .master("local[*]") \
    .getOrCreate()

print(f"Spark Version: {spark.version} - Session Started Successfully!")

# =====================================================================
# STEP 2: CREATE SPARK DATAFRAMES
# =====================================================================
print("\n--- Creating PySpark DataFrames ---")

# Employee Dataset
emp_data = [
    (101, "Aarav", "IT", 85000, "Mumbai"),
    (102, "Ananya", "HR", 62000, "Bangalore"),
    (103, "Rohan", "IT", 95000, "Delhi"),
    (104, "Isha", "Finance", 78000, "Mumbai"),
    (105, "Vikram", "IT", 88000, "Bangalore"),
    (106, "Neha", "Finance", 82000, "Delhi"),
    (107, "Priya", "HR", 58000, "Mumbai")
]
emp_columns = ["emp_id", "emp_name", "department", "salary", "location"]

df_emp = spark.createDataFrame(emp_data, schema=emp_columns)

# Department Info Dataset
dept_data = [
    ("IT", "Information Technology", "Floor 3"),
    ("HR", "Human Resources", "Floor 1"),
    ("Finance", "Financial Operations", "Floor 2"),
    ("Marketing", "Marketing & PR", "Floor 4")
]
dept_columns = ["department", "dept_full_name", "floor"]

df_dept = spark.createDataFrame(dept_data, schema=dept_columns)

print("Employee DataFrame Schema:")
df_emp.printSchema()

print("\nEmployee Data:")
df_emp.show()

# =====================================================================
# STEP 3: FILTERING & COLUMN TRANSFORMATIONS
# =====================================================================
print("\n--- Filtered Employees (Salary > 75000) ---")
df_high_salary = df_emp.filter(col("salary") > 75000)
df_high_salary.show()

print("\n--- Added Calculated Column: Bonus (10% of Salary) ---")
df_with_bonus = df_emp.withColumn("bonus", round(col("salary") * 0.10, 2))
df_with_bonus.show()

# =====================================================================
# STEP 4: GROUPING & AGGREGATIONS
# =====================================================================
print("\n--- Department-wise Aggregations ---")
df_dept_summary = df_emp.groupBy("department").agg(
    count("emp_id").alias("employee_count"),
    round(avg("salary"), 2).alias("average_salary"),
    sum("salary").alias("total_salary_budget")
)
df_dept_summary.show()

# =====================================================================
# STEP 5: PYSPARK JOINS (Inner & Left Joins)
# =====================================================================
print("\n--- Inner Join (Employee + Department Info) ---")
df_joined = df_emp.join(df_dept, on="department", how="inner")
df_joined.select("emp_id", "emp_name", "department", "dept_full_name", "salary", "floor").show()

# Stop Spark Session
spark.stop()
print("PySpark Execution Completed Successfully!")
```
